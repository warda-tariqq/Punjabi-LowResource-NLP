```python
# Upload and read corpus
with open("punjabi_shahmukhi_corpus_final.txt", "r", encoding="utf-8") as f:
    text = f.read()

# Basic statistics
tokens = text.split()
total_tokens = len(tokens)
vocab = set(tokens)
vocab_size = len(vocab)

print("Total Tokens:", total_tokens)
print("Vocabulary Size:", vocab_size)
```

```python
from collections import Counter

# Lexical Diversity (Type-Token Ratio)
ttr = vocab_size / total_tokens

print("Type-Token Ratio (TTR):", round(ttr, 4))

# Frequency analysis
freq = Counter(tokens)

print("\nTop 20 Most Frequent Words:\n")
for word, count in freq.most_common(20):
    print(word, ":", count)
```

```python
import re

# Sentence count
sentences = re.split(r'[.!؟]', text)
sentences = [s.strip() for s in sentences if s.strip()]

sentence_count = len(sentences)

# Average sentence length
avg_sentence_length = total_tokens / sentence_count

print("Sentence Count:", sentence_count)
print("Average Sentence Length:", round(avg_sentence_length,2))
```

```python
import re

# Better sentence splitting for Punjabi/Urdu/Shahmukhi
sentences = re.split(r'[۔!?؟\n]+', text)

# remove empty strings
sentences = [s.strip() for s in sentences if s.strip()]

sentence_count = len(sentences)

avg_sentence_length = total_tokens / sentence_count

print("Sentence Count:", sentence_count)
print("Average Sentence Length:", round(avg_sentence_length,2))
```

```python
# Show first 10 detected sentences
for i, s in enumerate(sentences[:10]):
    print(f"Sentence {i+1}:")
    print(s[:300])   # first 300 characters
    print("\n----------------\n")
```

```python
# Average word length
avg_word_length = sum(len(token) for token in tokens) / total_tokens

print("Average Word Length:", round(avg_word_length, 2))

# Top 50 most frequent words
print("\nTop 50 Most Frequent Words:\n")
for word, count in freq.most_common(50):
    print(word, ":", count)
```

```python
from collections import Counter

# Hapax legomena (words appearing once)
hapax = sum(1 for word, count in freq.items() if count == 1)

# Percentage of hapax
hapax_percent = (hapax / vocab_size) * 100

print("Hapax Legomena:", hapax)
print("Hapax Percentage:", round(hapax_percent,2), "%")
```

```python
import matplotlib.pyplot as plt

# Zipf distribution
counts = [count for word, count in freq.most_common(100)]

plt.figure(figsize=(8,5))
plt.plot(range(1,101), counts)
plt.xlabel("Word Rank")
plt.ylabel("Frequency")
plt.title("Zipf Distribution of Punjabi Shahmukhi Corpus")
plt.grid(True)
plt.show()
```

```python
from collections import Counter
import re

with open("punjabi_gurmukhi_corpus_cleaned.txt", "r", encoding="utf-8") as f:
    g_text = f.read()

g_tokens = g_text.split()
g_total = len(g_tokens)
g_vocab = len(set(g_tokens))
g_ttr = g_vocab / g_total
g_freq = Counter(g_tokens)

g_hapax = sum(1 for w,c in g_freq.items() if c == 1)
g_hapax_percent = (g_hapax / g_vocab) * 100

print("Gurmukhi Tokens:", g_total)
print("Gurmukhi Vocabulary:", g_vocab)
print("Gurmukhi TTR:", round(g_ttr,4))
print("Gurmukhi Hapax:", g_hapax)
print("Gurmukhi Hapax %:", round(g_hapax_percent,2))
```

```python
# Top 20 Gurmukhi words
print("\nTop 20 Gurmukhi Words:\n")
for word, count in g_freq.most_common(20):
    print(word, ":", count)
```

```python
# Vocabulary overlap
sh_vocab = set(tokens)
gu_vocab = set(g_tokens)

common_vocab = sh_vocab.intersection(gu_vocab)

print("Common Vocabulary:", len(common_vocab))
print("Overlap Percentage (Shahmukhi):", round(len(common_vocab)/len(sh_vocab)*100,2))
print("Overlap Percentage (Gurmukhi):", round(len(common_vocab)/len(gu_vocab)*100,2))
```

```python
# Show sample vocab
print("Shahmukhi sample:")
print(list(sh_vocab)[:20])

print("\nGurmukhi sample:")
print(list(gu_vocab)[:20])
```
